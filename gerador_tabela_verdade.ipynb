# Geração de Tabela Verdade

Este notebook tem como objetivo gerar a tabela verdade do Sistema Inteligente de Prioridade Energética e exportá-la automaticamente para um arquivo Excel formatado com cores institucionais e alinhamentos adequados.

## 1. Importação das Bibliotecas
Primeiro, importamos os módulos necessários para que o código funcione corretamente:

* **`itertools`**: Biblioteca nativa do Python, usada aqui para gerar todas as combinações possíveis de zeros e uns (produto cartesiano) para as entradas.
* **`pandas`**: Biblioteca poderosa para manipulação e estruturação de dados em formato de tabela (DataFrames).
* **`openpyxl`**: Biblioteca especializada em criar, ler e formatar planilhas do Excel (`.xlsx`). Dela importamos funções específicas para modificar fontes (`Font`), fundo das células (`PatternFill`) e alinhamento (`Alignment`).

```python
import itertools
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
```

## 2. Função: Geração da Tabela Verdade
A função `gerar_tabela_verdade()` é o núcleo lógico do script.

**Motivo:** Criar a tabela verdade de forma automatizada garante que não haverá erros humanos na hora de combinar os valores. Ela aplica a regra de negócio simplificada: `S = D AND (A OR (B AND NOT C))`, iterando sobre as 16 combinações possíveis das variáveis (A, B, C e D) e guardando o resultado em um formato que pode ser exportado depois.

```python
def gerar_tabela_verdade():
    """
    Gera os dados da tabela verdade baseados na lógica booleana do sistema.
    Retorna um DataFrame do pandas com os resultados.
    """
    colunas = [
        'A (Energia Solar)', 
        'B (Bateria > 40%)', 
        'C (Horário de Pico)', 
        'D (Eqp. Solicitado)', 
        'S (Carga Prioritária)'
    ]
    
    dados = []
    combinacoes = list(itertools.product([0, 1], repeat=4))

    for A, B, C, D in combinacoes:
        # Lógica simplificada: S = D AND (A OR (B AND NOT C))
        S = int(D and (A or (B and not C)))
        dados.append([A, B, C, D, S])

    return pd.DataFrame(dados, columns=colunas)
```

## 3. Função: Exportação e Formatação no Excel
A função `exportar_excel_formatado()` pega os dados estruturados e os transforma em um documento visualmente atraente e profissional.

**Motivo:** Entregar os dados brutos pode ser confuso. Esta função salva a tabela base e, em seguida, acessa as células diretamente para alterar o cabeçalho (adicionando fundo azul escuro e texto em negrito branco), além de centralizar tudo e alargar as colunas de forma inteligente para que o texto não fique cortado. Isso elimina a necessidade de formatar o Excel manualmente toda vez.

```python
def exportar_excel_formatado(df, nome_arquivo):
    """
    Exporta o DataFrame para Excel e aplica formatações visuais 
    (cores, alinhamento e ajuste de colunas).
    """
    try:
        df.to_excel(nome_arquivo, index=False, engine='openpyxl')

        wb = load_workbook(nome_arquivo)
        ws = wb.active

        estilo_cabecalho_fonte = Font(bold=True, color="FFFFFF") # Texto branco e negrito
        estilo_cabecalho_fundo = PatternFill(start_color="203764", end_color="203764", fill_type="solid") # Azul escuro
        estilo_alinhamento = Alignment(horizontal="center", vertical="center") # Centralizado

        for col in ws.columns:
            letra_coluna = col[0].column_letter
            tamanho_maximo = 0

            for celula in col:
                celula.alignment = estilo_alinhamento

                if celula.row == 1:
                    celula.font = estilo_cabecalho_fonte
                    celula.fill = estilo_cabecalho_fundo

                if celula.value:
                    tamanho_texto = len(str(celula.value))
                    if tamanho_texto > tamanho_maximo:
                        tamanho_maximo = tamanho_texto

            ws.column_dimensions[letra_coluna].width = tamanho_maximo + 4

        wb.save(nome_arquivo)
        print(f"Sucesso! Tabela exportada e formatada no arquivo: '{nome_arquivo}'.")

    except Exception as e:
        print(f"Ocorreu um erro ao gerar o arquivo Excel: {e}")
```

## 4. Execução do Script
Este é o ponto de entrada do script. É aqui que chamamos as funções definidas acima em sequência.

```python
if __name__ == "__main__":
    print("Gerando dados da Tabela Verdade...")
    df_tabela = gerar_tabela_verdade()
    
    arquivo_saida = 'Tabela_Verdade_GoodWe_Formatada.xlsx'
    exportar_excel_formatado(df_tabela, arquivo_saida)
    
    display(df_tabela)
```

## 5. Conclusão
Este script demonstra como a **lógica booleana** pode ser traduzida em código computacional para simular o comportamento de um sistema de hardware (Arquitetura de Computadores).

A automação da Tabela Verdade utilizando **Python e Pandas** evita falhas humanas e serve de prova de conceito para validar a expressão lógica otimizada: `S = D · (A + B · C̅)`. 

Além disso, a exportação automatizada e estilizada para Excel garante um documento final polido e pronto para ser anexado ao **Relatório Técnico da Sprint**, atestando o funcionamento das condições energéticas do sistema GoodWe proposto.
